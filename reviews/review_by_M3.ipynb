## M3 Analysis Module Review

For this part of the group project, my responsibility is to develop the analysis module, `data_analysis_M3.py`.

The main purpose of my module is to take the collated answer data produced by M2 and turn it into useful statistical and visual evidence. In particular, I focused on calculating the mean answer value for each question and producing visualisations that could help the group investigate whether the answer sequence shows any clear pattern.
